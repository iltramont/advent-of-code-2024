```python
#!/usr/bin/env python3
"""
Unisce più file PDF in un unico documento.

Uso:
    python unisci_pdf.py output.pdf file1.pdf file2.pdf file3.pdf
"""

import sys
from pathlib import Path

from pypdf import PdfWriter
```

```python
def unisci_pdf(output: str, input_files: list[str]) -> None:
    writer = PdfWriter()

    for percorso in input_files:
        file = Path(percorso)
        if not file.is_file():
            print(f"Errore: file non trovato -> {percorso}", file=sys.stderr)
            sys.exit(1)
        if file.suffix.lower() != ".pdf":
            print(f"Errore: non è un PDF -> {percorso}", file=sys.stderr)
            sys.exit(1)

        writer.append(str(file))
        print(f"Aggiunto: {file.name}")

    with open(output, "wb") as f:
        writer.write(f)

    print(f"\nCreato: {output}")
```
